```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import make_moons
from sklearn.cluster import KMeans
from sklearn.metrics import adjusted_rand_score
from sklearn.metrics import v_measure_score
from sklearn.metrics import silhouette_score
from sklearn.metrics import calinski_harabasz_score

x, y = make_moons(n_samples=500, noise=0.2, random_state=1)
```

1.) Visualize the original data

```python
plt.figure(figsize=(8,6))

plt.scatter(x[y == 0, 0], x[y == 0, 1], label = 'Class 0', alpha = 0.6)
plt.scatter(x[y == 1, 0], x[y == 1, 1], label = 'Class 1', alpha = 0.6)

plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.grid()
plt.axis('equal')
plt.show()
```

2.) Implement K-Means

We have two clusters, so we can initialize K = 2

```python
kmeans = KMeans(n_clusters=2, random_state=1, n_init=10)
kmeans.fit(x)
y_pred = kmeans.predict(x)
centroids = kmeans.cluster_centers_
print("Centroids Positions:", centroids)
```

3.) Visualize the predicted clusters

```python
plt.figure(figsize=(8, 6))

plt.scatter(x[y_pred == 0, 0], x[y_pred == 0, 1], label="Cluster 0", alpha=0.6)
plt.scatter(x[y_pred == 1, 0], x[y_pred == 1, 1], label="Cluster 1", alpha=0.6)
plt.scatter(centroids[:, 0], centroids[:, 1], color="black", marker="X", s=200, label="Centroids")

plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("K-means Clustering of Half-Moon Dataset")
plt.legend()
plt.grid()
plt.axis("equal")
plt.show()
```

4.) Evaluate the model

```python
# Adjusted Rand Index (ARI)

ari = adjusted_rand_score(y, y_pred)
print("Adjusted Rand Index:", ari)

# V-measure

v_measure = v_measure_score(y, y_pred)
print("\nV-measure:", v_measure)

# Silhouette Score

silhouette = silhouette_score(x, y_pred)
print("\nSilhouette Score:", silhouette)

# Calinski-Harabasz Index

ch = calinski_harabasz_score(x, y_pred)
print("\nCalinski-Harabasz Index:", ch)
```

Adjusted Rand Index: Decently poor agreemenet with the true clusters

V-measure: Poor homogeneity and completeness overall

Silhouette Score: Decent geometruc seperation

Alternate methods may be better like DBSCAN to be able to seperate this data better.
